### Resolve remaining missing values
Recalculate missing BMI only when height and weight are positive and the derived BMI is within the existing screening threshold. Preserve unknown blood type as a separate category; do not infer a biological attribute.

```python
bmi_missing = df['bmi'].isna()
calculated_bmi = df['weight'] / (df['height'] / 100)**2
valid_bmi = bmi_missing & df['height'].gt(0) & df['weight'].gt(0) & calculated_bmi.gt(0) & calculated_bmi.le(80)
print('BMI recovered:', valid_bmi.sum())
df.loc[valid_bmi, 'bmi'] = calculated_bmi[valid_bmi]
print('BMI still missing:', df['bmi'].isna().sum())
df['blood_type'] = df['blood_type'].fillna('unknown')
print('Blood type unknown:', df['blood_type'].eq('unknown').sum())
```

## Statistical validation
Assess selection effects and categorical association using the revised dataset.

```python
from scipy.stats import chi2_contingency
submitted = pd.read_csv('data/D1-processed.csv')
audit = pd.DataFrame([(len(t), t['covid19_positive'].replace({'True':1,'False':0}).astype(int).mean()) for t in [D1, submitted, df]], index=['Original','Submitted','Revised'], columns=['Rows','Positive rate'])
display(audit)
table = pd.crosstab(df['smoking'], df['covid19_positive'])
chi2, p_value, _, expected = chi2_contingency(table)
print('Chi-square:', chi2, 'p:', p_value, 'Cramer V:', np.sqrt(chi2/(table.values.sum()*min(table.shape[0]-1,table.shape[1]-1))))
print('Expected counts below 5:', (expected < 5).sum())
```
